```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.read_csv("Procrastination.csv")
```

```python
data = data.dropna()
```

```python
print(data.columns.tolist())
```

```python
target_col = "Do you consider yourself a procrastinator?"
y = data[target_col]
X = data.drop(columns=[target_col])

# Split the dataset into train and test sets with the corrected closing parenthesis
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

# Re-split the dataset with the encoded features
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, random_state=42
)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Now make predictions on the test set
y_pred = model.predict(X_test)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("AI-Based Procrastination Pattern Detection")
print("--------------------------------------------")
print("Algorithm: Random Forest")
print("Accuracy:", round(accuracy * 100, 2), "%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
